```python
class BTNode:
  def __init__(self, elem):
    self.elem = elem
    self.right = None
    self.left = None
```

```python
def inorder(root):
  if root == None:
    return

  inorder(root.left)
  print(root.elem, end = ' ')
  inorder(root.right)
```

```python
def tree_construction(arr, i = 1):
  if i>=len(arr) or arr[i] == None:
    return None
  p = BTNode(arr[i])
  p.left = tree_construction(arr, 2*i)
  p.right = tree_construction(arr, 2*i+1)
  return p


root2 = tree_construction([None, 'A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', None, None, None, 'I', 'J', None, 'k'])
inorder(root2)
```

```python
def insert(H, key):
    size = len(H)
    H.append(key)  # Append key to the end of the array
    swim(H, size)  # Call swim function with the newly added key's index

def swim(H, index):
    if index <= 1:
        return
    else:
        parent_index = index // 2
        if H[parent_index - 1] > H[index - 1]:  # Adjusting for 0-indexed arrays
            return
        else:
            # Swap parent and child if parent is less than child
            H[parent_index - 1], H[index - 1] = H[index - 1], H[parent_index - 1]
            swim(H, parent_index)  # Recursively call swim with the parent index

# Example usage:
heap = []
insert(heap, 10)
insert(heap, 20)
insert(heap, 5)
insert(heap, 15)
print(heap)  # Output: [20, 15, 5, 10]

```

```python
def delete(H):
    if len(H) == 0:
        return
    else:
        H[0], H[-1] = H[-1], H[0]  # Exchange root with last element
        H.pop()  # Remove the last element (previously the root)
        max_heapify(H, 0)  # Re-heapify the heap

def max_heapify(H, index):
    heap_size = len(H)
    left = 2 * index + 1
    right = 2 * index + 2

    largest = index

    if left < heap_size and H[left] > H[largest]:
        largest = left

    if right < heap_size and H[right] > H[largest]:
        largest = right

    if largest != index:
        H[index], H[largest] = H[largest], H[index]  # Swap elements
        max_heapify(H, largest)  # Recursively heapify the affected subtree

# Example usage:
heap = [30, 20, 10, 5, 15]
delete(heap)
print(heap)  # Output: [20, 15, 10, 5]
```
