```python
#list
#task01
d = {}
l = []
string = ''
count = 0
while string != 'STOP':
  string = input()
  if string != 'STOP':
    l.append(string)
print(l)
for i in l:
  key = i
  for k in l:
    if i == k:
      count += 1
    d[key] = count
  count = 0
for k, v in d.items():
  print(k, '-', v, 'times')
```

```python
#list1 main
x = ''
l1 = []
l2 = []
l3 = []
sum = 0
while x != 'STOP':
  x = input()
  l1.append(x)
for i in l1:
  if i not in l2 and i != 'STOP':
    l2.append(i)
for i in l2:
  for j in l1:
    if i == j:
      sum += 1
  l3.append(sum)
  sum = 0
for i in range(len(l3)):
  print(f'{l2[i]} - {l3[i]} times')
```

```python
#dictionary
#task01 ajaira
dict1 = {'a': 100, 'b':100, 'c':200,'d':300}
dict2 ={'a': 300, 'b':200, 'd':400, 'e':200}
dic3={}
unique_values = []
dict3 = dict1.copy()
dict3.update(dict2)
for key in dict2:
    if key in dict1:
        dict3[key] = dict2 [key] + dict1[key]
print(dict3)
for i in dic3.values():
  if i not in unique_values:
    unique_values.append(i)
print(unique_values)
for j in range(len(unique_values)-1):
  for i in range(len(unique_values)-1):
    if unique_values[i] > unique_values[i+1]:
      temp = unique_values[i]
      unique_values[i] = unique_values[i+1]
      unique_values[i+1] = temp
    else:
      continue  
print('Values: ',tuple(unique_values))
    
    
```

```python
#task 01 main
a=input().split(", ")
b=input().split(", ")
d1={}
d2={}
d3={}
lst=[]
for i in a:
    c,d=i.split(": ")
    d1[c]=int(d)
for j in b:
    c,d=j.split(": ")
    d2[c]=int(d)
print(d1)
print(d2)
for k in d1:
    count=0
    if k in d2:
        count+=d1[k]+d2[k]
        d3[k]=count
    else:
        d3[k]=d1[k]
for l in d2:
    if l not in d1:
        d3[l]=d2[l]
print(d3)
for v in d3.values():
    if v not in lst:
        lst.append(v)
lst.sort()
print("Values :",tuple(lst))
```
