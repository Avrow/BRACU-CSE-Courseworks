```python
#   PART - 1
def manhattan_distance(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def find_neighbors(cell, maze, total_rows, total_cols):
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    neighbors = []
    row, col = cell
    i = 0
    while i < len(directions):
        dr, dc = directions[i]
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < total_rows and 0 <= new_col < total_cols:
            if maze[new_row][new_col] == '0':
                neighbors.append((new_row, new_col))
        i += 1

    return neighbors

def build_path(came_from, start, end):
    steps = []
    current = end

    move_to_char = {
        (-1, 0): 'U',
        (1, 0): 'D',
        (0, -1): 'L',
        (0, 1): 'R'
    }

    while current != start:
        prev = came_from[current]
        move = (current[0] - prev[0], current[1] - prev[1])
        steps.append(move_to_char[move])
        current = prev

    return steps[::-1]

def insert_with_priority(queue, item):
    queue.append(item)
    queue.sort()

def pop_lowest_priority(queue):
    return queue.pop(0)

def solve_maze(input_data):
    total_rows, total_cols = map(int, input_data[0].split())
    start_row, start_col = map(int, input_data[1].split())
    goal_row, goal_col = map(int, input_data[2].split())

    maze_grid = []
    for i in range(3, 3 + total_rows):
      row = input_data[i].strip()
      maze_grid.append(row)
    start = (start_row, start_col)
    goal = (goal_row, goal_col)

    open_list = []
    insert_with_priority(open_list, (0, start))

    path_tracker = {}
    cost_from_start = {start: 0}
    total_cost = {start: manhattan_distance(start, goal)}  # f(n) = g + h

    visited = set()  #  re-checking nodes


    while open_list:
        _, current_cell = pop_lowest_priority(open_list)
        if current_cell in visited:
            continue
        visited.add(current_cell)
        if current_cell == goal:
            path = build_path(path_tracker, start, goal)
            return len(path), '->'.join(path)

        possible_moves = find_neighbors(current_cell, maze_grid, total_rows, total_cols)

        for neighbor in possible_moves:
            if neighbor in visited:
                continue

            new_cost_to_neighbor = cost_from_start[current_cell] + 1

            if neighbor not in cost_from_start or new_cost_to_neighbor < cost_from_start[neighbor]:
                cost_from_start[neighbor] = new_cost_to_neighbor
                estimated_total = new_cost_to_neighbor + manhattan_distance(neighbor, goal)
                total_cost[neighbor] = estimated_total
                path_tracker[neighbor] = current_cell
                insert_with_priority(open_list, (estimated_total, neighbor))
    return "1", ""
```

```python
given_maze = [
    "10 12",
    "0 1",
    "9 8",
    "#0##########",
    "#000000###0#",
    "###0###0##0#",
    "#0#0#000##0#",
    "#0#0###0##0#",
    "#0000#00##0#",
    "###0###0##0#",
    "#000#0000#0#",
    "#0#####0000#",
    "#######0####"
]

cost, path = solve_maze(given_maze)
print("Total Step :",cost)
print(path)
```

```python
#PART 2

def bfs_distance(gr, st, end):
    queue = [(st, 0)]   #fifo
    queue: (node, distance)
    visited = set([st]) # block repeat
    index = 0
    while index < len(queue):
         node, dist = queue[index]
         index += 1
         if node == end:
             return dist
         for neighbor in gr[node]:
             if neighbor not in visited:
                 visited.add(neighbor)
                 queue.append((neighbor, dist + 1))

    return "NO PATH" #0


def check_heuristics(input_lines):
    total_nodes, total_edges = map(int, input_lines[0].split())
    st_node, end_node = map(int, input_lines[1].split())
    heuristics = {}
    for i in range(2, 2 + total_nodes):
        node, value = map(int, input_lines[i].split()) # string,list
        heuristics[node] = value
    gr = {} #graph
    for i in range(1, total_nodes + 1):
        gr[i] = []

    for i in range(2 + total_nodes, 2 + total_nodes + total_edges): #1->2
        u, v = map(int, input_lines[i].split())                     #5->4,2,6,3
        gr[u].append(v)
        gr[v].append(u)
    not_ok = []
    for i in range(1, total_nodes + 1):
        real = bfs_distance(gr, i, end_node)
        estimate = heuristics[i]
        if estimate > real:
            not_ok.append(i)
    if not not_ok:
        return "1"
    else:
        text = ', '.join(str(x) for x in not_ok[:-1])
        text += f" and {not_ok[-1]}"
        return f"0\nHere nodes {text} are inadmissible."


def solve_admissibility_check(input_list):
    return check_heuristics(input_list)
```
