```python
#task 02
def fibonacci(x):
    a=0
    b=1
    total=0
    print(a,end=" ")
    print(b,end=" ")
    while total<x:
        total=a+b
        if total<x:
            print(total,end=" ")
            a=b
            b=total
fibonacci(10)
```

```python
#task2 extra
def fibo(x):
    a=0
    b=1
    print(a,end=" ")
    print(b,end=" ")
    for i in range (x-1):
        c=a+b
        a=b
        b=c
        print(c,end=" ")
fibo(10)
    
```

```python
#task 07
def show_palindrome(x):
    s=""
    s1=""

    for i in range(1,x+1):
        s+=str(i)
    for i in range (x-1,0,-1):
        s1+=str(i)
    print(s+s1,end=" ")
show_palindrome(5)


```

```python
#task08 main

def show_palidrome (n):
    s=""
    for i in range(1,n+1): 
        s+= str(i)+" "
    for i in range (n-1,0,-1):
         s+= str(i)+" "
    return s
def show_palidrome_triangle(n):
    for i in range (1,n+1):
         print("  "* (n-i)+show_palidrome (i)) 
show_palidrome_triangle(5)
```

```python
#task 08
def show_palindromic_triangle(x):
    for i in range (1,x+1):
        print(i,end="  ")
    for i in range(x-1,0,-1):
        print(i,end="  ")
def show_palindromic_triangle(h):
    j=1
    for i in range (h-1,-1,-1):
        print(" "*i,end="   ")
        show_palindrome(j)
        print()
        j+=1
show_palindromic_triangle(5)
```

```python
#task11
def rem_duplicate(x):
    a=x
    a=list(a)
    b=[]
    for i in a:
        if i not in b:
            b.append(i)
    b=tuple(b)
    return(b)
print(rem_duplicate((1,1,1,2,3,4,5,6,6,6,6,4,0,0,0)))
```

```python
#t20
def cacc(n,i,g,p):
    bonus=g*(p/100*i)
    if g>30:
        bonus+=10000
    elif g >=20:
        bonus+=5000

    print(n,"earned a bonus of", bonus , "taka for", g, "goals")

cacc("Neymar", 1200000, 35, 5)
```
