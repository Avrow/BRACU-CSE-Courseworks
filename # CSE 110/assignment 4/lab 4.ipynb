```python
#task10
string = input("Enter the strings: ")
for i in range(0, len(string)):
    z = ord(string[i])
    if z == 44:
        c = i
first = string[0:c]
second = string[c + 2::]
j = len(first)
m = len(second)
if j > m:
    for k in range(0, j):
        print(first[k], end="")
        if k < m:
            print(second[k], end="")
else:
    for k in range(0, m):
        if k < j:
            print(first[k], end="")
        print(second[k], end="")

```

```python
#task 10
x=input()
n1=0
i1=0
while x[n1]!=',':
    n1+=1
i2=n1+2
n2=len(x)
res=''
while i1<n1 and i2<n2:
    res+=x[i1]+x[i2]
    i1+=1
    i2+=1
if i1<n1:
    res+=x[i1:n1]
else:
    res+=x[i2:n2]
print(res)
```

```python
#task 1o main
inp=input()
c1=0
for i in inp:  
    if i==", ":
        break
    else:
        c1+=1
c2=c1+2
first_part=inp[0:c1]
sec_part=inp[c2:len(inp)]
ans=""
max=len(first_part)
if len(first_part)<len(sec_part):
    max=len(sec_part)
for i in range(0,max,1):
    if i<len(first_part):
         ans+=first_part[i]
    if i<len(sec_part):
        ans+=sec_part[i]
print(ans)
```

```python
#task10 ajaira
s1=str(input("Give me a word:"))
s2=str(input("Give me a second word:"))
ans=""
if len(s1)<=len(s2):
    num=len(s1)
else:
    num=len(s2)
for i in range(num):
    ans+=s1[i]
    ans+=s2[i]
print(ans)
if len(s1)<len(s2):
    ans+=s2[num :len(s2)]
if len(s2)<len(s1):
    ans+=s1[num:len(s1)]
print(ans)
```
